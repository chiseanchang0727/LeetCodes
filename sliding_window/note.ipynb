- how to know if sliding window fits the problem?
    - If the subarray matches the constraints, then any other subarray contains this array also matches the constraints.
    - If the subarray violates the constrains, then any other subarray contains this array also violates the constraints.
- create a array or dictionary for counting the alphabet appear times
    - `arr[0] = 1` stands for `a` appears 1 times.
    - `arr[1] = 2` stands for `b` appears 2 times.
- There are $n^2$ substring, and we check each them with $n$ times, the complexity is $O(n^3)$
- If we used sliding window, the complexity can be reduced to $O(n)$
    - because the bracket that slice the substring only move forward, which means the alphabet in the given string will be go through only one times.
